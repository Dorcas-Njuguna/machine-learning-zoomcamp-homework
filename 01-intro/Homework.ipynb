```python
# Step 1: Calculate the initial median
initial_median = df['horsepower'].median()

# Step 2: Calculate the most frequent value (mode)
most_frequent_value = df['horsepower'].mode()[0]

# Step 3: Fill missing values with the most frequent value
df['horsepower'] = df['horsepower'].fillna(most_frequent_value)

# Step 4: Calculate the median again
new_median = df['horsepower'].median()

# Step 5: Print results to see if it changed
print(f"Initial Median: {initial_median}")
print(f"Most Frequent Value (Mode): {most_frequent_value}")
print(f"New Median: {new_median}")

if new_median > initial_median:
    print("Answer: Yes, it increased")
elif new_median < initial_median:
    print("Answer: Yes, it decreased")
else:
    print("Answer: No")
```

```python
# 1. Select all cars from Asia
asia_cars = df[df['origin'] == 'Asia']
asia_cars
```

```python
# 2. Select only columns vehicle_weight and model_year, and get the first 7 values
X_df = asia_cars[['vehicle_weight', 'model_year']].head(7)
X_df
```

```python
# 3. Get the underlying NumPy array
X = X_df.to_numpy()
X
```

```python
# 4. Compute XTX (Matrix multiplication of X transpose and X) 
XTX = X.T @ X
XTX
```

```python
# 5.invert XTX
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
# 6. Create array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
# 7. Multiply the inverse of XTX with the transpose of X, and then by y to get w
w = XTX_inv @ X.T @ y
print("Weights (w):", w)
```

```python
# 8. What's the sum of all the elements of the result?
total_sum = w.sum()
print(f"Sum of all elements: {total_sum}")
```
